## **Question 1 :**  Soit $X \in \mathbb{R}^{n \times p}$ la matrice des données.
La normalisation consiste à centrer et réduire chaque variable :
$$ X_{norm} = \frac{X - \mu}{\sigma} $$

où $\mu$ est la moyenne des colonnes, et $\sigma$ l'écart-type.

```python
# (n, p)  =  (nombre_de_fromages, nombre_de_variables)
X = df.values  # conversion DataFrame → numpy array
#X
```

```python
X = df.select_dtypes(include=[np.number]).to_numpy()
#X
```

```python

# Moyenne colonne par colonne
mu = np.mean(X, axis=0)
#mu
```

```python
# Écart-type colonne par colonne
sigma = np.std(X, axis=0)
#sigma
```

```python
# Normalisation
Xnorm = (X - mu) / sigma
#Xnorm
```

##  Question 2 : Matrice de covariance


$$\text{M} = \frac{1}{n-1} X^T X $$



```python
# Pour la Question 2, on doit calculer la matrice de covariance à partir de la matrice X (déjà centrée, ou normalisée comme tu l’as fait juste avant).
#La matrice de covariance mesure les liens entre les variables.
```

```python
# M est une matrice carré (p × p).
# rowvar=False => colonnes sont des variables
M = np.corrcoef(Xnorm, rowvar=False)
M
```

##  Question 3 : Valeurs propres et vecteurs propres


$$ \text{M} v = \lambda v $$



```python
import numpy as np

#eigvals = variances expliquées par chaque composante,
#eigvecs = directions des composantes principales.
# M est la matrice de covariance (p x p)
# Décomposition en valeurs/vecteurs propres
eigvals, eigvecs = np.linalg.eig(M)
#eigvecs
```

```python
eigvals
```

```python
# Tri des composantes principales (du plus important au moins important)
# Tri des valeurs propres en ordre décroissant
idx = np.argsort(eigvals)[::-1]
eigvals = eigvals[idx]
eigvecs = eigvecs[:, idx]
eigvals
```

## Question 4 : Projection sur les composantes principales

$$ X_{proj} = X_{norm} \cdot V_k $$

avec $V_k$ les $k$ premiers vecteurs propres.



```python
# On sélectionne les deux premières composantes
# sélection des k premières composantes
# k : nombre de composantes principales
k= 2
V2 = eigvecs[:, :k]
# matrice p x 2 contenant les deux premières composantes
# projection des données sur 2 dimensions
Xproj2 = Xnorm @ V2
#Xproj2
```
