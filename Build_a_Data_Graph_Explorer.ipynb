```python
import os
import requests
import pandas as pd


# Get a .csv file in putting the url in the code
data_dir = 'data'
btc_file_path = os.path.join(data_dir, 'btc-market-price.csv')
btc_url = 'https://raw.githubusercontent.com/ine-rmotr-curriculum/freecodecamp-intro-to-pandas/master/data/btc-market-price.csv'

if not os.path.exists(data_dir):
    os.makedirs(data_dir)
    print(f'Created directory : {data_dir}')

# Download the file it's up-to-date and not corrupted
print(f"Downloading {btc_file_path}...")
response = requests.get(btc_url)
if response.status_code == 200:
  with open(btc_file_path, 'wb') as f:
    f.write(response.content)
  print("Download complete.")
else:
  # If download fails, raise an error
  raise FileNotFoundError(f"Failed to download file from {btc_url}. Status code: {response.status_code}")

df = pd.read_csv(btc_file_path, names=['Timestamp', 'Price'], parse_dates=True) # index_col=0,
df.head(2)
columns_names = df.columns
# print(columns_names.dtype)
list_names = columns_names.tolist()
print(list_names)

```
